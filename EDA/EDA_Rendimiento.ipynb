## **Tratamiento del conjunto de datos:**

Se realiza la carga del Dataframe (df) para visualizar y entender la estructura de los datos y su interpretabilidad. Se evidencia que la mayoría de las variables se encuentran codificadas, por lo anterior, se realizó un proceso de homologación de variables categóricas reemplazando los valores numéricos codificados por sus respectivas etiquetas descriptivas.

Esta transformación permite mejorar la interpretabilidad del conjunto de datos, facilitar el análisis estadístico y asegurar una lectura más clara de los resultados en los análisis exploratorios y visualizaciones, manteniendo al mismo tiempo la equivalencia de la información original.

Las variables a homologar son las siguientes:

| Variable | Homologación |
|---|---|
| Gender | 0 → Masculino<br>1 → Femenino |
| Ethnicity | 0 → Caucasian<br>1 → African American<br>2 → Asian<br>3 → Other |
| ParentalEducation | 0 → None<br>1 → High School<br>2 → Some College<br>3 → Bachelor's<br>4 → Higher |
| Tutoring | 0 → No<br>1 → Sí |
| ParentalSupport | 0 → None<br>1 → Low<br>2 → Moderate<br>3 → High<br>4 → Very High |
| Extracurricular | 0 → No<br>1 → Sí |
| Sports | 0 → No<br>1 → Sí |
| Music | 0 → No<br>1 → Sí |
| Volunteering | 0 → No<br>1 → Sí |
| GradeClass | 0 → A<br>1 → B<br>2 → C<br>3 → D<br>4 → F |

Además, se evidencia con la función info que no hay datos nulos, duplicados, por lo tanto no se requiere hacer limpieza o ajuste adicional al df.

```python
# Se crea nuevo dataframe para no alterar el original y se reemplazan los valores numéricos por sus correspondientes categorías para facilitar la interpretación de los datos.
datos = df.copy()

# 1. Gender
datos["Gender"] = datos["Gender"].replace({
    0: "Male",
    1: "Female"
})

# 2. Ethnicity
datos["Ethnicity"] = datos["Ethnicity"].replace({
    0: "Caucasian",
    1: "African American",
    2: "Asian",
    3: "Other"
})

# 3. ParentalEducation
datos["ParentalEducation"] = datos["ParentalEducation"].replace({
    0: "None",
    1: "High School",
    2: "Some College",
    3: "Bachelor's",
    4: "Higher"
})

# 4. Tutoring
datos["Tutoring"] = datos["Tutoring"].replace({
    0: "No",
    1: "Yes"
})

# 5. ParentalSupport
datos["ParentalSupport"] = datos["ParentalSupport"].replace({
    0: "None",
    1: "Low",
    2: "Moderate",
    3: "High",
    4: "Very High"
})

# 6. Extracurricular
datos["Extracurricular"] = datos["Extracurricular"].replace({
    0: "No",
    1: "Yes"
})

# 7. Sports
datos["Sports"] = datos["Sports"].replace({
    0: "No",
    1: "Yes"
})

# 8. Music
datos["Music"] = datos["Music"].replace({
    0: "No",
    1: "Yes"
})

# 9. Volunteering
datos["Volunteering"] = datos["Volunteering"].replace({
    0: "No",
    1: "Yes"
})

# 10. GradeClass
datos["GradeClass"] = datos["GradeClass"].replace({
    0: "A",
    1: "B",
    2: "C",
    3: "D",
    4: "F"
})

# Se verifica el resultado
datos.head(10)
```

## **Feature Engineering**

Se realiza el feature engineering, creando una nueva variable significativa: Total activities.

**Total_activities**

Esta variable agrupa la participación en actividades extracurriculares (columnas Extracurricular, Sports, Music, Volunteering).Lo que nos permitirá evaluar si la participación en actividades adicionales influye positiva o negativamente en el desempeño académico.


```python
datos_fe = datos.copy()

datos_fe["Total_activities"] = (
    (datos_fe["Extracurricular"] == "Yes").astype(int) +
    (datos_fe["Sports"] == "Yes").astype(int) +
    (datos_fe["Music"] == "Yes").astype(int) +
    (datos_fe["Volunteering"] == "Yes").astype(int)
)

datos_fe.head()
```

```python
datos_fe["Total_activities"].value_counts().sort_index()
```

```python
datos_fe.groupby("Total_activities")["GPA"].mean().round(2)
```

**Análisis variable Total_activities**  

La mayoría de los estudiantes participa en una o ninguna actividad extracurricular, siendo muy pocos los que tienen una alta carga de actividades.

En cuanto al rendimiento académico, se observa que el GPA tiende a mejorar a medida que aumenta la participación en actividades, pasando de 1.75 en estudiantes sin actividades a valores cercanos o superiores a 2.0 en quienes participan en dos o más.

Sin embargo, en niveles muy altos de participación no se evidencia una mejora significativa adicional, lo que sugiere que un nivel moderado de actividades podría ser más beneficioso.

En conclusión, participar en actividades extracurriculares parece tener un impacto positivo en el rendimiento académico, especialmente cuando se mantiene un equilibrio.

Se procede a realizar una tabla dinámica, en la que se analiza el promedio académico (GPA) cruzando dos variables: la cantidad de actividades extracurriculares (Total_activities) y el nivel de apoyo de los padres (ParentalSupport).

```python
pivot_gpa = pd.pivot_table(
    datos_fe,
    values="GPA",
    index="Total_activities",
    columns="ParentalSupport",
    aggfunc="mean"
).round(2)

orden = ["None", "Low", "Moderate", "High", "Very High"]
pivot_gpa = pivot_gpa.reindex(columns=orden)

display(pivot_gpa)
```

El apoyo de los padres parece ser un factor determinante en el rendimiento del estudiante, independientemente de cuántas actividades realice. Se puede observar que hay una tendencia clara: a mayor apoyo, mejor promedio académico.

Los niveles de apoyo parental "Very High" y "High" suelen tener los promedios más altos (superando casi siempre el 2.0). Por otro lado, los niveles "None" y "Low" presentan los promedios más bajos. Sin embargo, como hallazgo curioso se evidencia que los estudiantes con apoyo "Low" o "Moderate" que realizan 4 actividades tienen promedios bastante altos, 2.44 y 2.42, respectivamente, superando incluso a los de apoyo High.

## **Ranking por actividades extra**

```python
# Reestructuramos el DataFrame de formato ancho a formato largo, con el fin de que todas las actividades estén en una sola columna 

actividades = ["Extracurricular", "Sports", "Music", "Volunteering"]

long_acts = datos_fe.melt(
    id_vars="GPA",
    value_vars=actividades,
    var_name="Actividad",
    value_name="Participa"
)

# Sacamos el GPA promedio solo para quienes participan

impacto_actividad = (
    long_acts[long_acts["Participa"] == "Yes"]
    .groupby("Actividad")["GPA"]
    .mean()
    .reset_index()
)

# Realizamos el ranking de actividades

impacto_actividad["Ranking"] = (
    impacto_actividad["GPA"]
    .rank(ascending=False, method="dense")
)

impacto_actividad.sort_values("Ranking").reset_index(drop=True)
```

El ranking evidencia que, entre las actividades analizadas, la participación en Music se asocia con el mayor GPA promedio, seguida por Extracurricular y Sports, mientras que Volunteering presenta el GPA promedio más bajo. Esto sugiere que las actividades con mayor componente estructurado o cognitivo tienden a relacionarse con un mejor desempeño académico, aunque todas contribuyen en distinta medida al rendimiento estudiantil.

## **Segmentación por género de estudiantes con mejor promedio académico**

```python
# Filtramos estudiantes con GradeClass A y B (rango de mejor GPA)

df_seg = datos_fe[datos_fe["GradeClass"].isin(["A", "B"])].copy()

# Definimos las condiciones y segmentos que queremos para nuestra segmentación

conditions = [
    df_seg["Gender"] == "Female",
    df_seg["Gender"] == "Male"
]

choices = [
    "Mujeres sobresalientes (A) o buen desempeño (B)",
    "Hombres sobresalientes (A) o buen desempeño (B)"
]

# Creamos la variable de segmentación

df_seg["Segmento_estudiante"] = np.select(
    conditions,
    choices,
    default="Otros alto desempeño"
)

# Resumen

segment_summary = (
    df_seg
    .groupby("Segmento_estudiante")
    .agg(
        estudiantes=("StudentID", "count"),
        GPA_promedio=("GPA", "mean"),
        estudio_promedio=("StudyTimeWeekly", "mean"),
        ausencias_promedio=("Absences", "mean")
    )
    .sort_values("GPA_promedio", ascending=False)
)

display(segment_summary.round(2))
```

Las mujeres con calificación A o B presentan un GPA promedio ligeramente superior y menos ausencias en comparación con los hombres del mismo nivel académico, a pesar de que ambos grupos dedican un tiempo de estudio semanal muy similar. Esto sugiere que, dentro del alto desempeño académico, existen diferencias de género asociadas principalmente a la asistencia y no al esfuerzo de estudio, lo cual puede influir en el rendimiento final.

## **Perfil académico y hábitos de estudio de estudiantes con alto desempeño y participación extracurricular**

```python
actividades = ["Extracurricular", "Sports", "Music", "Volunteering"]

#  Buscamos la actividad que más practican

actividad_mas_practicada = (
    datos_fe[actividades]
    .eq("Yes")
    .sum()
    .idxmax()
)

actividad_mas_practicada
```

```python
datos_fe[actividad_mas_practicada].value_counts()
```

```python
# Filtramos con query (Mujeres, GradeClass A o B, Participan en la actividad más practicada)

casos_interesantes = datos_fe[
    (datos_fe["Gender"] == "Female") &
    (datos_fe[actividad_mas_practicada] == "Yes") &
    (datos_fe["GradeClass"].isin(["A", "B"]))
][[
    "StudentID", "Age", "Gender", "GradeClass",
    "GPA", "StudyTimeWeekly", "Absences",
    actividad_mas_practicada
]]

display(casos_interesantes.head(10))
```

```python
# Filtramos con query (Hombres, GradeClass A o B, Participan en la actividad más practicada)

casos_interesantes = datos_fe[
    (datos_fe["Gender"] == "Male") &
    (datos_fe[actividad_mas_practicada] == "Yes") &
    (datos_fe["GradeClass"].isin(["A", "B"]))
][[
    "StudentID", "Age", "Gender", "GradeClass",
    "GPA", "StudyTimeWeekly", "Absences",
    actividad_mas_practicada
]]

display(casos_interesantes.head(10))
```

Esta segmentación refleja que la participación extracurricular, combinada con buenos hábitos académicos generales, se asocia con un alto rendimiento, aunque las estrategias y niveles de esfuerzo varían significativamente entre los estudiantes, tanto para hombres como mujeres.

A pesar de este buen rendimiento (GradeClass A y B), se observa una alta variabilidad en los hábitos de estudio: mientras algunas estudiantes dedican más de 18 horas semanales, otras presentan tiempos de estudio considerablemente bajos (menos de 3 horas) y aun así mantienen GPA superiores a 3.0. Esto sugiere que el rendimiento no depende únicamente del tiempo de estudio, sino también de otros factores como la eficiencia del aprendizaje o el contexto académico.

En cuanto a la asistencia, también se observa que estudiantes con altas ausencias pueden tener un buen desempeño.

## **Top**

```python
top_combinations = (
    datos_fe
    .groupby(
        ["Gender", "GradeClass", "ParentalSupport", "Total_activities"],
        observed=False
    )
    .agg(
        estudiantes=("StudentID", "count"),
        GPA_promedio=("GPA", "mean"),
        ausencias_promedio=("Absences", "mean"),
        estudio_promedio=("StudyTimeWeekly", "mean")
    )
    .reset_index()
)

# 

top_combinations = top_combinations[
    top_combinations["estudiantes"] >= 15
]

#


top_combinations = top_combinations.sort_values(
    ["GPA_promedio", "estudiantes"],
    ascending=[False, False]
)

display(top_combinations.head(10).round(3))
```

El top de combinaciones muestra que los estudiantes con calificación B y niveles moderados o altos de apoyo parental tienen los GPA promedio más altos, especialmente cuando participan en al menos una actividad extracurricular. A medida que disminuye la calificación (nivel C), se observa una tendencia de reducción del GPA, acompañada de mayores ausencias y, en general, menor tiempo de estudio, lo que indica que el apoyo parental y la participación extracurricular pueden ser factores que ayudan a mejorar el rendimiento académico.
